## 4.2 Import library

```python
import re
from collections import defaultdict
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print('Library IR siap digunakan.')
```

## 4.3 Corpus dan preprocessing

Corpus berisi dokumen contoh. Fungsi tokenisasi mengubah teks menjadi lowercase dan mengambil token alfanumerik; token hasilnya ditampilkan agar proses preprocessing dapat diperiksa.

```python
documents = {
    1: 'Python digunakan untuk analisis data dan machine learning.',
    2: 'Information retrieval mencari dokumen yang relevan dengan query.',
    3: 'Machine learning dapat digunakan untuk klasifikasi teks.',
    4: 'Search engine menggunakan indexing dan ranking dokumen.',
    5: 'TF-IDF memberikan bobot pada kata yang penting dalam dokumen.',
    6: 'Inverted index menyimpan daftar dokumen untuk setiap term.',
    7: 'Semantic search menggunakan embedding untuk memahami makna.',
    8: 'Natural language processing memproses data berupa teks.',
    9: 'Cosine similarity membandingkan kemiripan vektor dokumen.',
    10: 'BM25 dapat digunakan untuk ranking pada sistem retrieval.',
}

def tokenize(text):
    return re.findall(r'[a-z0-9]+', text.lower())

tokenized = {doc_id: tokenize(text) for doc_id, text in documents.items()}
pd.DataFrame({'document_id': documents.keys(), 'text': documents.values(), 'tokens': tokenized.values()})
```

## 4.4 Pembuatan inverted index

Index memetakan term ke postings list document ID. Setiap term disimpan satu kali per dokumen.

```python
def build_inverted_index(documents):
    index = defaultdict(set)
    for doc_id, text in documents.items():
        for term in set(tokenize(text)):
            index[term].add(doc_id)
    return {term: sorted(doc_ids) for term, doc_ids in index.items()}

index = build_inverted_index(documents)
index_table = pd.DataFrame({
    'term': list(index.keys()),
    'postings': list(index.values()),
}).sort_values('term').reset_index(drop=True)
display(index_table.head(15))
print('Jumlah term dalam vocabulary:', len(index))
```

### Contoh normalisasi dan postings list

Kapitalisasi diseragamkan, tanda baca menjadi pemisah, dan pengulangan term dalam dokumen yang sama hanya menghasilkan satu document ID di postings list.

```python
contoh_teks = 'Index, INDEX! index'
contoh_token = tokenize(contoh_teks)
contoh_index = build_inverted_index({101: contoh_teks, 102: 'Index retrieval'})

print('Token hasil normalisasi:', contoh_token)
print('Posting untuk term index:', contoh_index['index'])
assert contoh_token == ['index', 'index', 'index']
assert contoh_index['index'] == [101, 102]
```

## 4.5 Boolean retrieval

AND mengambil irisan postings list, OR menggabungkannya, dan NOT mengurangkan postings term dari seluruh koleksi.

```python
def boolean_and(index, terms):
    sets = [set(index.get(term, [])) for term in terms]
    return set.intersection(*sets) if sets else set()

def boolean_or(index, terms):
    result = set()
    for term in terms:
        result.update(index.get(term, []))
    return result

def boolean_not(index, term, universe):
    return set(universe) - set(index.get(term, []))

print('machine AND learning:', sorted(boolean_and(index, ['machine', 'learning'])))
print('retrieval OR indexing:', sorted(boolean_or(index, ['retrieval', 'indexing'])))
print('NOT classification:', sorted(boolean_not(index, 'classification', documents)))
```

## 4.6 Ranked retrieval dengan TF-IDF

Vectorizer dipasang pada corpus dan query digunakan untuk mengurutkan dokumen berdasarkan cosine similarity.

```python
texts = [documents[doc_id] for doc_id in sorted(documents)]
vectorizer = TfidfVectorizer(tokenizer=tokenize, token_pattern=None)
X = vectorizer.fit_transform(texts)

def rank_query(query, top_k=5):
    query_vector = vectorizer.transform([query])
    scores = cosine_similarity(query_vector, X).ravel()
    order = np.argsort(scores)[::-1][:top_k]
    return pd.DataFrame({
        'document_id': [sorted(documents)[i] for i in order],
        'score': scores[order],
        'text': [texts[i] for i in order],
    })

query = 'information retrieval'
ranking = rank_query(query)
print('Ranking untuk query:', query)
display(ranking.style.format({'score': '{:.4f}'}))
```

## 4.7 Evaluasi precision dan recall

Relevance set di bawah merupakan label contoh yang ditentukan manual untuk demonstrasi. Precision@5 dan recall@5 dihitung untuk tiap query; evaluasi berbasis relevance judgment dan metrik ranking dibahas dalam literatur IR [4], [6], [7].

```python
relevant = {
    'machine learning': {1, 3},
    'information retrieval': {2, 4, 6, 10},
    'semantic search': {7, 9},
}

def precision_recall_at_k(retrieved, gold, k):
    actual = set(retrieved[:k])
    hits = len(actual & gold)
    precision = hits / k if k else 0.0
    recall = hits / len(gold) if gold else 0.0
    return precision, recall

evaluation = []
for query_text, gold in relevant.items():
    retrieved = rank_query(query_text, top_k=5)['document_id'].tolist()
    precision, recall = precision_recall_at_k(retrieved, gold, 5)
    evaluation.append({
        'query': query_text, 'retrieved': retrieved,
        'precision@5': precision, 'recall@5': recall,
    })

evaluation_table = pd.DataFrame(evaluation)
display(evaluation_table)
```

```python
assert X.shape[0] == len(documents)
assert len(index) > 0
assert evaluation_table['precision@5'].between(0, 1).all()
assert evaluation_table['recall@5'].between(0, 1).all()
print('Validasi IR: PASS')
```

## 4.8 Analisis otomatis hasil eksperimen

Script berikut merangkum ukuran corpus dan index, contoh Boolean retrieval, ranking aktual, serta nilai evaluasi. Narasi dan angka dihasilkan dari variabel eksperimen.

```python
# Analisis otomatis BAB 4: penjelasan berdasarkan hasil eksperimen aktual
num_docs = len(documents)
vocab_size = len(index)
posting_count = sum(len(postings) for postings in index.values())
avg_postings = posting_count / vocab_size if vocab_size else 0.0
first_result = ranking.iloc[0]
mean_precision = float(evaluation_table['precision@5'].mean())
mean_recall = float(evaluation_table['recall@5'].mean())
boolean_hits = sorted(boolean_and(index, ['machine', 'learning']))

print('ANALISIS OTOMATIS BAB 4')
print(f'Corpus memuat {num_docs} dokumen dan {vocab_size} term unik.')
print(f'Inverted index memiliki {posting_count} pasangan term-dokumen, dengan rata-rata {avg_postings:.2f} document ID per term.')
print(f'Pencarian Boolean machine AND learning menemukan dokumen {boolean_hits}.')
print(f'Untuk query "{query}", hasil teratas adalah dokumen {int(first_result["document_id"])} dengan cosine similarity {float(first_result["score"]):.4f}.')
print(f'Rata-rata precision@5 = {mean_precision:.3f}; rata-rata recall@5 = {mean_recall:.3f} dari {len(evaluation_table)} query evaluasi.')
if mean_precision > mean_recall:
    print('Interpretasi: pada label contoh ini, hasil top-5 relatif tepat, tetapi cakupan dokumen relevan masih lebih rendah.')
elif mean_recall > mean_precision:
    print('Interpretasi: pada label contoh ini, sistem mencakup lebih banyak dokumen relevan, dengan sebagian hasil top-5 yang tidak relevan.')
else:
    print('Interpretasi: precision dan recall rata-rata seimbang pada label contoh ini.')
print('Catatan: nilai evaluasi bergantung pada relevance set manual dan corpus kecil yang digunakan.')
```
