```python
import sys
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import cv2

print('Python:', sys.version.split()[0])
print('PyTorch:', torch.__version__)
print('NumPy:', np.__version__)
print('OpenCV:', cv2.__version__)
print('CUDA available:', torch.cuda.is_available())
if torch.cuda.is_available():
    print('GPU:', torch.cuda.get_device_name(0))
```

## Latar Belakang dan Tujuan

Computer Vision memproses gambar sebagai tensor numerik. Environment yang konsisten penting agar eksperimen dapat direproduksi di laptop, VS Code, maupun Google Colab. Tugas ini memverifikasi NumPy, OpenCV, Matplotlib, dan PyTorch.

Tujuan khusus: memastikan library dapat diimpor, memeriksa ketersediaan GPU, dan membuktikan bahwa layer konvolusi menghasilkan tensor dengan bentuk yang benar.

## Verifikasi Conv2d

Dengan kernel 3×3 dan padding 1, ukuran spasial input dipertahankan. Jumlah channel output berubah sesuai jumlah filter, sedangkan dimensi batch tetap sama.

```python
x = torch.randn(2, 3, 224, 224)
conv = nn.Conv2d(3, 64, kernel_size=3, padding=1)
y = conv(x)
print('Input:', tuple(x.shape))
print('Output:', tuple(y.shape))
assert tuple(y.shape) == (2, 64, 224, 224)
print('VALIDATION: PASS')
```

## Penjelasan Hasil dan Kesimpulan

Output versi membantu mendeteksi konflik instalasi. `CUDA available=True` berarti PyTorch dapat memakai GPU, sedangkan `False` berarti proses tetap dapat berjalan di CPU. Bentuk input `(batch, channel, height, width)` adalah format standar PyTorch.

Environment berhasil diverifikasi jika seluruh import berjalan, versi library tercetak, dan validasi bentuk tensor menghasilkan `PASS`. Langkah berikutnya adalah memakai notebook P02 untuk pemrosesan citra dasar.

```python
# Kesimpulan otomatis
status = 'PASS' if tuple(y.shape) == (2, 64, 224, 224) else 'FAIL'
print('ANALISIS OTOMATIS')
print(f'- CUDA tersedia: {torch.cuda.is_available()}')
print(f'- Bentuk tensor: {tuple(x.shape)} -> {tuple(y.shape)}')
print('KESIMPULAN:', 'Environment dan Conv2d berhasil.' if status == 'PASS' else 'Periksa konfigurasi Conv2d.')
```
